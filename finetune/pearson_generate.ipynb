```python
import os
import json
import numpy as np
from scipy.stats import pearsonr
from utils import process_json

curDir = os.getcwd()
# modelName = "hyenadna-medium-160k-seqlen-hf"
modelName = "HGDNA"
resName = "housekeeping_generation_new.json"

resDir =  os.path.join(curDir, "result", modelName, resName)
data = process_json(resDir)
```

```python
all_activity = []
for k, v in data.items():
    activity = [_[-1] for _ in v['res'].values()]
    prompt = [int(k) for _ in range(len(activity))]
    activity = np.array([prompt, activity], dtype=float)
    all_activity.append(activity)

all_activity = np.concatenate(all_activity, axis=1)
corelation, pvalue = pearsonr(all_activity[0], all_activity[1])
print(f"Corelation: {corelation}, pvalue: {pvalue}")
```
